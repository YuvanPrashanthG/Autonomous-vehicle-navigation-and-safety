# Synthetic Data generation

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n_samples = 1000

# Generate synthetic data
time = np.linspace(0, 100, n_samples)  # Time in seconds
speed = np.random.uniform(20, 120, n_samples)  # Speed in km/h
acceleration = np.random.normal(0, 2, n_samples)  # Acceleration in m/s²
steering_angle = np.random.uniform(-10, 10, n_samples)  # Steering angle in degrees
distance_to_obstacle = 50 + speed * 0.5 + np.random.normal(0, 5, n_samples)  # Distance in meters
brake_intensity = np.clip(100 - distance_to_obstacle, 0, 100)  # Brake intensity as percentage

data = pd.DataFrame({
    "Time (s)": time,
    "Speed (km/h)": speed,
    "Acceleration (m/s^2)": acceleration,
    "Steering Angle (°)": steering_angle,
    "Distance to Obstacle (m)": distance_to_obstacle,
    "Brake Intensity (%)": brake_intensity
})

data.to_csv("synthetic_autonomous_vehicle_data.csv", index=False)

print("Synthetic dataset generated and saved as 'synthetic_autonomous_vehicle_data.csv'")
```

# Regression Model

```python

data = pd.read_csv("synthetic_autonomous_vehicle_data.csv")

X = data[["Speed (km/h)", "Acceleration (m/s^2)"]].values
y = data["Distance to Obstacle (m)"].values
X = np.c_[np.ones(X.shape[0]), X]

#Bulid Model
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Coefficients (Beta):", beta)
y_pred = X @ beta

# Evaluate the model
mse = np.mean((y - y_pred) ** 2)
ss_total = np.sum((y - np.mean(y)) ** 2)
ss_residual = np.sum((y - y_pred) ** 2)
r_squared = 1 - (ss_residual / ss_total)

print("Mean Squared Error (MSE):", mse)
print("R-squared:", r_squared)

# Display a few predictions
print("Actual values:", y[:5])
print("Predicted values:", y_pred[:5])
```
